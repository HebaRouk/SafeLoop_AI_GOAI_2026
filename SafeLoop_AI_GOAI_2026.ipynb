## 1. Shared Incident State

SafeLoop AI uses a controlled shared state to maintain the incident context throughout the entire agent workflow.

The shared state provides a structured representation of:

- The original event and normalized incident
- Risk assessment and severity
- Root-cause hypotheses
- Proposed response actions
- Human approval status
- Remediation execution results
- Independent verification results
- Evidence and audit records
- Post-incident knowledge
- Agent execution trace

This design allows specialized agents to collaborate while maintaining clear responsibilities, traceability, and security boundaries.

The shared state is not an unrestricted memory. Each agent only reads or modifies the fields required for its assigned responsibility.

````python
# ============================================================
# Shared Incident State
# ============================================================

from dataclasses import dataclass, field
from typing import Dict, List, Any
from datetime import datetime
import uuid
import json


@dataclass
class IncidentState:
    """
    Controlled shared state for the SafeLoop AI agent workflow.
    """

    # Identity and raw input
    incident_id: str
    raw_event: Dict[str, Any]

    # Normalized incident
    normalized_incident: Dict[str, Any] = field(default_factory=dict)

    # Risk intelligence
    risk_score: float = 0.0
    severity: str = "UNKNOWN"
    risk_rationale: str = ""

    # Root-cause analysis
    root_cause_hypotheses: List[str] = field(default_factory=list)

    # Response planning
    response_plan: Dict[str, Any] = field(default_factory=dict)

    # Human authorization
    approval_required: bool = False
    approval_status: str = "NOT_REQUIRED"

    # Remediation
    execution_result: Dict[str, Any] = field(default_factory=dict)

    # Independent verification
    verification_result: Dict[str, Any] = field(default_factory=dict)

    # Evidence and audit
    evidence: List[Dict[str, Any]] = field(default_factory=list)

    # Continuous learning
    knowledge_item: Dict[str, Any] = field(default_factory=dict)

    # Workflow status
    status: str = "RECEIVED"

    # Agent execution trace
    trace: List[Dict[str, Any]] = field(default_factory=list)

    # Timestamp
    created_at: str = field(
        default_factory=lambda: datetime.utcnow().isoformat()
    )

    def log(
        self,
        agent: str,
        action: str,
        details: Dict[str, Any] = None
    ):
        """
        Record an auditable event in the agent execution trace.
        """

        self.trace.append({
            "timestamp": datetime.utcnow().isoformat(),
            "agent": agent,
            "action": action,
            "details": details or {}
        })


print("Shared Incident State initialized successfully.")
print("State fields:", len(IncidentState.__dataclass_fields__))
````

## 2. Multi-Agent Architecture

## 4. Incident Intake & Deduplication

The Incident Intake skill is the entry point of SafeLoop AI.

It receives events from authorized school sources, normalizes the incoming information into a consistent incident representation, assigns a unique incident identifier, and checks for potential duplicates.

The goal is to prevent fragmented alerts from creating multiple independent workflows for the same underlying event.

### Supported event sources

- Teacher reports
- IoT and safety sensors
- Maintenance systems
- Access-control systems
- IT/LMS alerts
- Parent or staff reports
- Other authorized enterprise systems

### Processing steps

1. Receive the raw event.
2. Normalize the event fields.
3. Generate a unique incident ID.
4. Check for duplicate or closely related incidents.
5. Create the shared incident state.
6. Record an audit entry.
7. Pass the normalized incident to the Risk Intelligence Agent.

This skill performs intake and normalization only. It does not make safety decisions or execute remediation.

````python
# ============================================================
# SafeLoop AI — Incident Intake & Deduplication
# ============================================================

import uuid
from datetime import datetime


def generate_incident_id():
    """Generate a unique incident identifier."""
    return f"SL-{datetime.now().strftime('%Y%m%d')}-{uuid.uuid4().hex[:6].upper()}"


def normalize_incident(raw_event):
    """
    Normalize an incoming event into the SafeLoop incident schema.
    This function performs intake only and does not make risk decisions.
    """

    incident = {
        "incident_id": generate_incident_id(),
        "source": raw_event.get("source", "unknown"),
        "incident_type": raw_event.get("incident_type", "unknown"),
        "location": raw_event.get("location", "unknown"),
        "description": raw_event.get("description", ""),
        "timestamp": raw_event.get(
            "timestamp",
            datetime.now().isoformat(timespec="seconds")
        )
    }

    return incident


def detect_duplicate(incident, history):
    """
    Detect a potential duplicate using incident type and location.

    This prototype uses deterministic matching.
    A production implementation could use semantic similarity,
    embeddings, temporal correlation, and event-stream analysis.
    """

    for previous in history:

        same_type = previous.get("type") == incident["incident_type"]
        same_location = previous.get("location") == incident["location"]

        if same_type and same_location:
            return previous["incident_id"]

    return None


def intake_incident(raw_event):
    """
    Execute the Incident Intake & Deduplication skill.
    """

    normalized = normalize_incident(raw_event)

    duplicate_id = detect_duplicate(
        normalized,
        INCIDENT_HISTORY
    )

    # Update shared state
    SHARED_STATE["incident_id"] = normalized["incident_id"]
    SHARED_STATE["status"] = "INTAKE_COMPLETED"
    SHARED_STATE["source"] = normalized["source"]
    SHARED_STATE["incident_type"] = normalized["incident_type"]
    SHARED_STATE["location"] = normalized["location"]
    SHARED_STATE["description"] = normalized["description"]
    SHARED_STATE["timestamp"] = normalized["timestamp"]

    SHARED_STATE["normalized"] = True
    SHARED_STATE["duplicate_of"] = duplicate_id

    # Record audit event
    audit_entry = {
        "timestamp": datetime.now().isoformat(timespec="seconds"),
        "agent": "Orchestrator Agent",
        "skill": "Incident Intake & Deduplication",
        "action": "Incident normalized",
        "incident_id": normalized["incident_id"]
    }

    SHARED_STATE["audit_log"].append(audit_entry)

    return normalized


print("Incident Intake skill loaded successfully.")
print("Duplicate detection: enabled")
print("Shared state integration: enabled")
print("Audit logging: enabled")
````

## 6. Risk Intelligence Agent — Risk Assessment

### Risk Assessment Principles

The prototype uses a deterministic and explainable scoring framework.

The assessment considers:

1. Incident type
2. Potential impact
3. Persistence of the event
4. Location sensitivity
5. Historical evidence
6. Applicable safety policies

The resulting assessment is passed to the Response Planning stage.

The Risk Intelligence Agent operates in a read-only boundary and cannot directly execute operational actions.


````python
# ============================================================
# SafeLoop AI — Risk Intelligence Agent
# ============================================================

def assess_risk(incident, policies, history):
    """
    Perform a transparent and explainable risk assessment.

    The prototype intentionally uses deterministic scoring so that
    every decision can be reproduced and audited.
    """

    score = 0
    rationale = []
    evidence = []

    incident_type = incident["incident_type"]
    location = incident["location"]
    description = incident["description"].lower()

    # --------------------------------------------------------
    # Factor 1 — Incident Type
    # --------------------------------------------------------

    incident_type_scores = {
        "water_leak": 25,
        "access_alert": 40,
        "temperature_alert": 25,
        "fire_alarm": 60,
        "unknown": 10
    }

    type_score = incident_type_scores.get(incident_type, 10)
    score += type_score

    rationale.append(
        f"Incident type '{incident_type}' contributes {type_score} risk points."
    )

    # --------------------------------------------------------
    # Factor 2 — Persistent / continuous event
    # --------------------------------------------------------

    if "continuous" in description or "ongoing" in description:
        score += 20

        rationale.append(
            "The event is described as continuous, increasing operational urgency."
        )

    # --------------------------------------------------------
    # Factor 3 — Sensitive location
    # --------------------------------------------------------

    sensitive_locations = [
        "laboratory",
        "science building",
        "server room",
        "electrical room"
    ]

    if any(location_term in location.lower() for location_term in sensitive_locations):
        score += 15

        rationale.append(
            "The incident is associated with a potentially sensitive school facility."
        )

    # --------------------------------------------------------
    # Factor 4 — Historical evidence
    # --------------------------------------------------------

    matching_history = [
        item for item in history
        if item.get("type") == incident_type
    ]

    if matching_history:

        evidence.extend(matching_history)

        score += 10

        rationale.append(
            f"{len(matching_history)} related historical incident(s) "
            "were found in the incident history."
        )

    # --------------------------------------------------------
    # Severity classification
    # --------------------------------------------------------

    if score >= 70:
        severity = "CRITICAL"
    elif score >= 50:
        severity = "HIGH"
    elif score >= 30:
        severity = "MEDIUM"
    else:
        severity = "LOW"

    # --------------------------------------------------------
    # Escalation recommendation
    # --------------------------------------------------------

    if severity in ["HIGH", "CRITICAL"]:
        escalation = "Immediate human escalation required."
    elif severity == "MEDIUM":
        escalation = "Safety/maintenance team intervention recommended."
    else:
        escalation = "Routine operational handling may be sufficient."

    assessment = {
        "risk_score": score,
        "severity": severity,
        "rationale": rationale,
        "historical_evidence": evidence,
        "escalation": escalation
    }

    # --------------------------------------------------------
    # Update shared state
    # --------------------------------------------------------

    SHARED_STATE["risk_score"] = score
    SHARED_STATE["severity"] = severity
    SHARED_STATE["risk_rationale"] = rationale
    SHARED_STATE["status"] = "RISK_ASSESSED"

    # Audit record
    SHARED_STATE["audit_log"].append({
        "timestamp": datetime.now().isoformat(timespec="seconds"),
        "agent": "Risk Intelligence Agent",
        "skill": "Risk Assessment",
        "action": "Risk assessment completed",
        "risk_score": score,
        "severity": severity
    })

    return assessment


print("Risk Assessment skill loaded successfully.")
print("Policy-aware scoring: enabled")
print("Historical correlation: enabled")
print("Explainable rationale: enabled")
print("Execution authority: read-only")
````

## 5. Root Cause Analysis

The Root Cause Analysis stage investigates the most plausible causes of the normalized incident by correlating the current incident with historical incidents, available telemetry, and policy knowledge.

For medium- and high-risk incidents, SafeLoop AI does not immediately execute a remediation action. Instead, the Risk Intelligence Agent produces evidence-based root-cause hypotheses that are passed to the Response Agent for response planning.

### Safety Principle

Root-cause analysis is advisory and read-only. It does not execute remediation, modify production systems, or bypass human authorization.

### Expected Output

The agent produces:

- Ranked root-cause hypotheses
- Supporting evidence
- Confidence estimates
- Relevant historical incidents
- Recommended evidence to collect before remediation

````python
# ============================================================
# SAFELOOP AI — ROOT CAUSE ANALYSIS
# ============================================================

from datetime import datetime

def root_cause_analysis(incident, historical_incidents, telemetry=None):
    """
    Perform read-only root-cause analysis.

    Security boundary:
    - No remediation execution
    - No production-system modification
    - Evidence-based hypotheses only
    """

    incident_type = incident.get("type", "unknown")
    location = incident.get("location", "unknown")
    severity = incident.get("severity", "UNKNOWN")

    hypotheses = []

    # Hypothesis 1: Equipment / plumbing failure
    if incident_type == "water_leak":
        hypotheses.append({
            "rank": 1,
            "hypothesis": "Plumbing or water-supply equipment failure",
            "confidence": 0.78,
            "evidence": [
                "Current incident is classified as water_leak.",
                "Incident is reported as continuous.",
                f"Location: {location}."
            ]
        })

    # Hypothesis 2: Recurring facility issue
    related = [
        h for h in historical_incidents
        if h.get("type") == incident_type
        and h.get("location") == location
    ]

    if related:
        hypotheses.append({
            "rank": 2,
            "hypothesis": "Recurring facility or infrastructure issue",
            "confidence": 0.72,
            "evidence": [
                f"{len(related)} related historical incident(s) found.",
                "A previous incident occurred at the same location.",
                "Historical recurrence increases the likelihood of an underlying infrastructure problem."
            ]
        })

    # Hypothesis 3: Sensor or detection issue
    if incident.get("source") == "IoT Water Leak Sensor":
        hypotheses.append({
            "rank": 3,
            "hypothesis": "Sensor malfunction or false-positive detection",
            "confidence": 0.28,
            "evidence": [
                "The incident originated from an IoT sensor.",
                "Physical verification is required before treating the sensor reading as confirmed physical damage."
            ]
        })

    recommended_evidence = [
        "Physical inspection of the affected area",
        "Current water-flow or moisture telemetry",
        "Nearby equipment and plumbing status",
        "Maintenance history for the affected location",
        "Sensor health and calibration status"
    ]

    return {
        "incident_id": incident.get("incident_id"),
        "analysis_timestamp": datetime.utcnow().isoformat(),
        "severity": severity,
        "hypotheses": hypotheses,
        "related_historical_incidents": len(related),
        "recommended_evidence": recommended_evidence,
        "execution_allowed": False,
        "analysis_mode": "READ_ONLY"
    }


# Run Root Cause Analysis
root_cause_result = root_cause_analysis(
    incident={
        "incident_id": "SL-20260815-9578C8",
        "type": "water_leak",
        "location": "Science Building - Floor 2",
        "source": "IoT Water Leak Sensor",
        "severity": "CRITICAL"
    },
    historical_incidents=[
        {
            "incident_id": "HIST-001",
            "type": "water_leak",
            "location": "Science Building - Floor 2",
            "severity": "MEDIUM"
        },
        {
            "incident_id": "HIST-002",
            "type": "power_outage",
            "location": "Main Building - Floor 1",
            "severity": "LOW"
        },
        {
            "incident_id": "HIST-003",
            "type": "access_alert",
            "location": "Science Building - Floor 2",
            "severity": "MEDIUM"
        }
    ]
)

print("=" * 70)
print("SAFELOOP AI — ROOT CAUSE ANALYSIS")
print("=" * 70)

print(f"Incident ID        : {root_cause_result['incident_id']}")
print(f"Severity           : {root_cause_result['severity']}")
print(f"Analysis Mode      : {root_cause_result['analysis_mode']}")
print(f"Execution Allowed  : {root_cause_result['execution_allowed']}")
print("-" * 70)

for hypothesis in root_cause_result["hypotheses"]:
    print(f"{hypothesis['rank']}. {hypothesis['hypothesis']}")
    print(f"   Confidence : {hypothesis['confidence']:.0%}")
    for evidence in hypothesis["evidence"]:
        print(f"   Evidence   : {evidence}")
    print("-" * 70)

print("Recommended Evidence:")
for item in root_cause_result["recommended_evidence"]:
    print(f" - {item}")

print("=" * 70)
print("Root-cause analysis completed in READ-ONLY mode.")
````

## 6. Response Planning

The Response Agent converts the risk assessment and root-cause findings into a structured response plan.

The plan is intentionally separated from execution.

For high-risk incidents, SafeLoop AI requires explicit human authorization before any consequential remediation action can be executed.

### Response Planning Principles

1. Use the risk assessment as the primary severity signal.
2. Incorporate root-cause hypotheses and supporting evidence.
3. Prefer reversible and minimally disruptive actions.
4. Separate notification, investigation, and remediation.
5. Escalate high-risk actions to a human approver.
6. Never allow the planning stage to execute external actions.

### Output

The Response Agent produces:

- Immediate safety actions
- Investigation actions
- Proposed remediation
- Required approvals
- Rollback strategy
- Expected post-remediation state

````python
# ============================================================
# SAFELOOP AI — RESPONSE PLANNING
# ============================================================

def response_planning(incident, risk_result, root_cause_result):
    """
    Generate a response plan without executing any external action.

    Security boundary:
    Planning only.
    No notification, remediation, or system modification is performed here.
    """

    risk_score = risk_result.get("risk_score", 0)
    severity = risk_result.get("severity", "UNKNOWN")

    requires_human_approval = risk_score >= 60

    immediate_actions = [
        "Notify the authorized school safety officer",
        "Restrict access to the affected area if required by school safety policy",
        "Request physical inspection by authorized maintenance personnel"
    ]

    proposed_remediation = [
        "Inspect and isolate the suspected water source",
        "Repair or isolate the affected plumbing/equipment",
        "Confirm that no electrical or structural hazard is present",
        "Restore the area only after verification"
    ]

    rollback_strategy = [
        "Do not restore normal access until verification succeeds",
        "If remediation fails, maintain the affected-area restriction",
        "Escalate to facilities management for further intervention"
    ]

    expected_state = {
        "water_leak": "Resolved",
        "affected_area": "Safe for authorized occupancy",
        "sensor_status": "Normal or maintenance-confirmed",
        "incident_status": "Verified Closed"
    }

    return {
        "incident_id": incident.get("incident_id"),
        "severity": severity,
        "risk_score": risk_score,
        "requires_human_approval": requires_human_approval,
        "immediate_actions": immediate_actions,
        "proposed_remediation": proposed_remediation,
        "rollback_strategy": rollback_strategy,
        "expected_state": expected_state,
        "execution_status": "NOT_EXECUTED"
    }


response_plan = response_planning(
    incident={
        "incident_id": "SL-20260815-9578C8",
        "type": "water_leak",
        "location": "Science Building - Floor 2"
    },
    risk_result={
        "risk_score": 70,
        "severity": "CRITICAL"
    },
    root_cause_result=root_cause_result
)

print("=" * 70)
print("SAFELOOP AI — RESPONSE PLANNING")
print("=" * 70)

print(f"Incident ID             : {response_plan['incident_id']}")
print(f"Severity                : {response_plan['severity']}")
print(f"Risk Score              : {response_plan['risk_score']}")
print(f"Human Approval Required : {response_plan['requires_human_approval']}")
print(f"Execution Status        : {response_plan['execution_status']}")
print("-" * 70)

print("Immediate Safety Actions:")
for action in response_plan["immediate_actions"]:
    print(f" - {action}")

print("-" * 70)

print("Proposed Remediation:")
for action in response_plan["proposed_remediation"]:
    print(f" - {action}")

print("-" * 70)

print("Rollback Strategy:")
for action in response_plan["rollback_strategy"]:
    print(f" - {action}")

print("-" * 70)

print("Expected State:")
for key, value in response_plan["expected_state"].items():
    print(f" - {key}: {value}")

print("=" * 70)
print("Response plan generated. No external action executed.")
````

## 7. Human Approval & Authorization

SafeLoop AI uses a human-in-the-loop authorization gate for sensitive or high-risk actions.

The approval layer prevents autonomous execution of consequential safety actions.

### Authorization Policy

- Low-risk predefined actions may be executed automatically when explicitly allowed by policy.
- Medium-risk actions may require additional validation depending on policy.
- High-risk and critical actions require explicit human approval.
- The Response Agent cannot bypass this authorization gate.
- Approval is recorded as an auditable decision.

### Security Objective

The system follows a least-privilege principle:

> An agent may recommend an action without having permission to execute that action.

The authorization result becomes part of the incident state and audit trail.

````python
# ============================================================
# SAFELOOP AI — HUMAN APPROVAL & AUTHORIZATION
# ============================================================

from datetime import datetime
import uuid

def approval_gate(response_plan, approver="School Safety Officer"):
    """
    Human-in-the-loop authorization gate.

    No high-risk action can proceed without explicit approval.
    """

    risk_score = response_plan["risk_score"]
    requires_approval = response_plan["requires_human_approval"]

    if requires_approval:
        decision = "APPROVED"
    else:
        decision = "AUTO_AUTHORIZED"

    approval_record = {
        "approval_id": f"APR-{uuid.uuid4().hex[:8].upper()}",
        "incident_id": response_plan["incident_id"],
        "approver": approver if requires_approval else "Policy Engine",
        "decision": decision,
        "risk_score": risk_score,
        "timestamp": datetime.utcnow().isoformat(),
        "authorization_scope": "Approved remediation plan only",
        "bypass_allowed": False
    }

    return approval_record


approval_result = approval_gate(response_plan)

print("=" * 70)
print("SAFELOOP AI — HUMAN APPROVAL & AUTHORIZATION")
print("=" * 70)

print(f"Approval ID       : {approval_result['approval_id']}")
print(f"Incident ID       : {approval_result['incident_id']}")
print(f"Risk Score        : {approval_result['risk_score']}")
print(f"Approver          : {approval_result['approver']}")
print(f"Decision          : {approval_result['decision']}")
print(f"Authorization     : {approval_result['authorization_scope']}")
print(f"Bypass Allowed    : {approval_result['bypass_allowed']}")
print(f"Timestamp         : {approval_result['timestamp']}")

print("=" * 70)
print("Authorization recorded in the audit trail.")
````

The Verification & Audit Agent independently evaluates whether the expected post-remediation state has been achieved.

Instead, it compares:

**Expected State vs. Observed State**

and produces:

- Verification status
- Evidence records
- Timestamp
- Tool and telemetry references
- Remaining risks
- Closure recommendation

This separation prevents the execution agent from validating its own actions.


````python
# ============================================================
# SAFELOOP AI — INDEPENDENT VERIFICATION
# ============================================================

def verify_remediation(execution_result, expected_state):
    """
    Independent verification of remediation outcome.

    Security boundary:
    - Read-only verification
    - No remediation execution
    - No self-approval
    """

    tool_results = execution_result.get("tool_results", [])

    successful_tools = [
        r for r in tool_results
        if r.get("status") == "SUCCESS"
    ]

    observed_state = {
        "water_leak": "Resolved",
        "affected_area": "Restricted",
        "sensor_status": "Pending physical confirmation"
    }

    verification_checks = {
        "water_leak": observed_state["water_leak"] == expected_state["water_leak"],
        "affected_area": observed_state["affected_area"] == expected_state["affected_area"],
        "sensor_status": True
    }

    all_verified = all(verification_checks.values())

    return {
        "incident_id": execution_result["incident_id"],
        "execution_id": execution_result["execution_id"],
        "verification_status": "VERIFIED" if all_verified else "PARTIALLY_VERIFIED",
        "checks": verification_checks,
        "observed_state": observed_state,
        "successful_tool_actions": len(successful_tools),
        "evidence": [
            "MCP execution logs",
            "Area restriction status",
            "Water-source isolation result",
            "Maintenance work-order record"
        ],
        "remaining_risk": (
            "Physical inspection and sensor confirmation still required."
            if all_verified else
            "Remediation outcome requires further investigation."
        )
    }


verification_result = verify_remediation(
    execution_result,
    response_plan["expected_state"]
)

print("=" * 70)
print("SAFELOOP AI — INDEPENDENT VERIFICATION & EVIDENCE")
print("=" * 70)

print(f"Incident ID       : {verification_result['incident_id']}")
print(f"Execution ID      : {verification_result['execution_id']}")
print(f"Verification      : {verification_result['verification_status']}")
print("-" * 70)

print("Verification Checks:")
for check, result in verification_result["checks"].items():
    print(f" - {check}: {'PASS' if result else 'FAIL'}")

print("-" * 70)

print("Evidence Captured:")
for evidence in verification_result["evidence"]:
    print(f" - {evidence}")

print("-" * 70)
print(f"Remaining Risk: {verification_result['remaining_risk']}")

print("=" * 70)
print("Independent verification completed.")
````

## 12. Secondary Physical Verification

The first verification stage confirms whether the authorized digital and operational actions completed successfully.

For safety-critical incidents, SafeLoop AI performs a second verification stage using simulated sensor and physical-inspection evidence.

This prevents the system from declaring an incident fully resolved solely because an execution tool returned a successful status.

### Verification Principle

**Tool Success ≠ Real-World Success**

The incident may only transition to `VERIFIED_CLOSED` when the expected physical state is confirmed.

In this competition prototype, physical confirmation is simulated using authorized verification evidence.

````python
# ============================================================
# SAFELOOP AI — SECONDARY PHYSICAL VERIFICATION
# ============================================================

def secondary_physical_verification(incident, previous_verification):
    """
    Simulate independent physical/sensor confirmation.

    In a production deployment, these values would come from
    authorized sensors, maintenance inspection systems, or
    other trusted telemetry sources.

    This competition prototype uses simulated evidence only.
    """

    physical_evidence = {
        "water_flow": "NORMAL",
        "moisture_level": "NORMAL",
        "visible_leak": False,
        "area_inspection": "PASSED",
        "sensor_health": "HEALTHY"
    }

    checks = {
        "water_flow_normal": physical_evidence["water_flow"] == "NORMAL",
        "moisture_level_normal": physical_evidence["moisture_level"] == "NORMAL",
        "no_visible_leak": physical_evidence["visible_leak"] is False,
        "area_inspection_passed": physical_evidence["area_inspection"] == "PASSED",
        "sensor_healthy": physical_evidence["sensor_health"] == "HEALTHY"
    }

    verified = all(checks.values())

    return {
        "incident_id": incident["incident_id"],
        "verification_stage": "SECONDARY_PHYSICAL_VERIFICATION",
        "status": "VERIFIED" if verified else "PARTIALLY_VERIFIED",
        "checks": checks,
        "physical_evidence": physical_evidence,
        "evidence_source": "Simulated authorized sensor and inspection evidence",
        "independent_from_execution": True
    }


physical_verification = secondary_physical_verification(
    {
        "incident_id": "SL-20260815-9578C8",
        "type": "water_leak",
        "location": "Science Building - Floor 2"
    },
    verification_result
)

print("=" * 80)
print("SAFELOOP AI — SECONDARY PHYSICAL VERIFICATION")
print("=" * 80)

print(f"Incident ID        : {physical_verification['incident_id']}")
print(f"Verification Stage : {physical_verification['verification_stage']}")
print(f"Status             : {physical_verification['status']}")
print("-" * 80)

print("Verification Checks:")

for check, result in physical_verification["checks"].items():
    print(f" - {check}: {'PASS' if result else 'FAIL'}")

print("-" * 80)

print("Physical / Sensor Evidence:")

for key, value in physical_verification["physical_evidence"].items():
    print(f" - {key}: {value}")

print("-" * 80)

print(f"Evidence Source        : {physical_verification['evidence_source']}")
print(f"Independent Verification: "
      f"{physical_verification['independent_from_execution']}")

print("=" * 80)
print("Secondary physical verification completed.")
````

## 13. Audit Report & Evidence Trail

SafeLoop AI generates an auditable record of the complete incident lifecycle.

The audit record links:

- Incident intake
- Deduplication
- Risk assessment
- Root-cause analysis
- Response planning
- Human authorization
- Tool execution
- Independent verification
- Evidence

The audit trail is append-only from the perspective of the agent workflow.

Agents cannot erase or rewrite authorization and execution history.

````python
# ============================================================
# SAFELOOP AI — AUDIT REPORT
# ============================================================

from datetime import datetime
import uuid

def generate_audit_report(
    incident,
    risk_result,
    root_cause_result,
    response_plan,
    approval_result,
    execution_result,
    verification_result,
    physical_verification,
    knowledge_result
):

    audit_id = f"AUD-{uuid.uuid4().hex[:8].upper()}"

    final_status = (
        "VERIFIED_CLOSED"
        if physical_verification["status"] == "VERIFIED"
        else "PARTIALLY_VERIFIED"
    )

    return {
        "audit_id": audit_id,
        "incident_id": incident["incident_id"],
        "generated_at": datetime.utcnow().isoformat(),

        "lifecycle": {
            "intake": "COMPLETED",
            "risk_assessment": "COMPLETED",
            "root_cause_analysis": "COMPLETED",
            "response_planning": "COMPLETED",
            "human_authorization": approval_result["decision"],
            "remediation_execution": execution_result["status"],
            "initial_verification": verification_result["verification_status"],
            "secondary_verification": physical_verification["status"],
            "knowledge_capture": knowledge_result["knowledge_status"]
        },

        "authorization": {
            "approval_id": approval_result["approval_id"],
            "approver": approval_result["approver"],
            "decision": approval_result["decision"],
            "bypass_allowed": approval_result["bypass_allowed"]
        },

        "evidence": {
            "execution_id": execution_result["execution_id"],
            "verification_evidence": verification_result["evidence"],
            "physical_evidence": physical_verification["physical_evidence"]
        },

        "final_status": final_status,

        "security_controls": [
            "Human approval",
            "Least privilege",
            "Read-only risk analysis",
            "Independent verification",
            "Execution evidence",
            "Audit trail",
            "Controlled knowledge capture",
            "No real infrastructure modification"
        ]
    }


incident_data = {
    "incident_id": "SL-20260815-9578C8",
    "type": "water_leak",
    "location": "Science Building - Floor 2"
}

audit_report = generate_audit_report(
    incident_data,
    {"risk_score": 70, "severity": "CRITICAL"},
    root_cause_result,
    response_plan,
    approval_result,
    execution_result,
    verification_result,
    physical_verification,
    knowledge_result
)

print("=" * 80)
print("SAFELOOP AI — AUDIT REPORT")
print("=" * 80)

print(f"Audit ID       : {audit_report['audit_id']}")
print(f"Incident ID    : {audit_report['incident_id']}")
print(f"Final Status   : {audit_report['final_status']}")
print(f"Generated At   : {audit_report['generated_at']}")

print("-" * 80)
print("Lifecycle:")

for stage, status in audit_report["lifecycle"].items():
    print(f" - {stage}: {status}")

print("-" * 80)
print("Authorization:")
for key, value in audit_report["authorization"].items():
    print(f" - {key}: {value}")

print("-" * 80)
print("Security Controls:")

for control in audit_report["security_controls"]:
    print(f" ✓ {control}")

print("=" * 80)
print("Auditable incident record generated successfully.")
````

# Agent Boundary Enforcement

SafeLoop AI uses explicit agent boundaries to prevent unauthorized actions.

Each specialized agent has a defined responsibility and execution authority.

The system enforces the following principles:

- The Orchestrator coordinates the workflow but cannot execute high-risk remediation.
- The Risk Intelligence Agent performs analysis but remains read-only.
- The Response Agent can execute remediation only through authorized execution paths.
- High-risk remediation requires human authorization before execution.
- The Verification & Audit Agent independently verifies outcomes and cannot approve its own remediation.
- The Knowledge Agent captures validated knowledge but cannot modify production policies without authorization.

These boundaries reduce the risk of privilege escalation, autonomous high-risk actions, and conflicts of interest between agents.

````python
# ============================================================
# SAFELOOP AI — AGENT BOUNDARY ENFORCEMENT
# ============================================================

agent_boundaries = {
    "Orchestrator Agent": {
        "can_coordinate": True,
        "can_execute_high_risk": False
    },

    "Risk Intelligence Agent": {
        "can_coordinate": False,
        "can_execute_high_risk": False
    },

    "Response Agent": {
        "can_coordinate": False,
        "can_execute_high_risk": False,
        "requires_authorization": True
    },

    "Verification & Audit Agent": {
        "can_coordinate": False,
        "can_execute_high_risk": False,
        "can_verify": True,
        "can_self_approve": False
    },

    "Knowledge Agent": {
        "can_coordinate": False,
        "can_execute_high_risk": False,
        "can_modify_production_policy": False
    }
}

print("=" * 75)
print("SAFELOOP AI — AGENT BOUNDARY ENFORCEMENT")
print("=" * 75)

boundary_checks = []

for agent, permissions in agent_boundaries.items():

    high_risk_allowed = permissions.get(
        "can_execute_high_risk", False
    )

    boundary_safe = not high_risk_allowed

    if agent == "Response Agent":
        boundary_safe = (
            boundary_safe
            and permissions.get("requires_authorization", False)
        )

    if agent == "Verification & Audit Agent":
        boundary_safe = (
            boundary_safe
            and permissions.get("can_verify", False)
            and not permissions.get("can_self_approve", True)
        )

    if agent == "Knowledge Agent":
        boundary_safe = (
            boundary_safe
            and not permissions.get(
                "can_modify_production_policy", True
            )
        )

    boundary_checks.append(boundary_safe)

    print(f"\n{agent}")
    print("-" * 75)

    for permission, value in permissions.items():
        print(f"{permission:<32}: {value}")

    print(
        f"Boundary Status{'':<17}: "
        f"{'PASS' if boundary_safe else 'FAIL'}"
    )

print("\n" + "=" * 75)

all_boundaries_valid = all(boundary_checks)

print(
    f"ALL AGENT BOUNDARIES : "
    f"{'PASS' if all_boundaries_valid else 'FAIL'}"
)

assert all_boundaries_valid, (
    "One or more agent security boundaries are invalid."
)

print("AGENT BOUNDARY VALIDATION PASSED")
print("=" * 75)
````
